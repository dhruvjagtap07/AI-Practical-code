```python
# Assignment 1: BFS & DFS

from collections import defaultdict, deque

class Graph:
    def __init__(self):
        self.graph = defaultdict(list)
        
    # Add edge (undirected)
    def add_edge(self, u, v):
        self.graph[u].append(v)
        self.graph[v].append(u)
        
# ---------------- DFS RECURSIVE ----------------
    def dfs_recursive(self, node, visited):
        visited.add(node)
        print(node, end=" ")
        for neighbor in self.graph[node]:
            if neighbor not in visited:
                self.dfs_recursive(neighbor, visited)
                
# ---------------- DFS ITERATIVE ----------------
    def dfs_iterative(self, start):
        visited = set()
        stack = [start]
        while stack:
            node = stack.pop()
            if node not in visited:
                print(node, end=" ")
                visited.add(node)
                
                for neighbor in reversed(self.graph[node]):
                    if neighbor not in visited:
                        stack.append(neighbor)
                    
# ---------------- BFS ITERATIVE ----------------
    def bfs_iterative(self, start):
        visited = set()
        queue = deque([start])
        visited.add(start)
        
        while queue:
            node = queue.popleft()
            print(node, end=" ")
            
            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
                    
# ---------------- BFS RECURSIVE ----------------
        
    def bfs_recursive(self, queue, visited):
        if not queue:
            return
        node = queue.popleft()
        print(node, end=" ")
        
        for neighbor in self.graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                self.bfs_recursive(queue, visited)
                
# ---------------- DRIVER CODE ----------------
g = Graph()

# Input
n = int(input("Enter number of edges: "))
print("Enter edges (u v):")

for _ in range(n):
    u, v = map(int, input().split())
    g.add_edge(u, v)
    
start = int(input("Enter starting node: "))

# DFS Recursive
print("\nDFS Recursive:")
visited = set()
g.dfs_recursive(start, visited)

# DFS Iterative
print("\nDFS Iterative:")
g.dfs_iterative(start)

# BFS Iterative
print("\nBFS Iterative:")
g.bfs_iterative(start)

# BFS Recursive
print("\nBFS Recursive:")
visited = set([start])
g.bfs_recursive(deque([start]), visited)
```

```python
# Assignment 2: A * Algorithm
import heapq

def heuristic(a, b):
    return abs(a[0]-b[0]) + abs(a[1]-b[1])

def a_star_algorithm(start, goal, grid):
    rows, cols = len(grid), len(grid[0])
    
    open_list = []
    heapq.heappush(open_list, (0, start))
    
    g_cost = {start: 0}
    parent = {start: None}
    
    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    
    while open_list:
        _, current = heapq.heappop(open_list)
        
        if current == goal:
            path = []
            while current:
                path.append(current)
                current = parent[current]
                
            return path[::-1]
        
        for d in directions:
            neighbor = (current[0] + d[0], current[1] + d[1])
            r, c = neighbor
            
            if 0 <= r < rows and 0 <= c < cols and grid[r][c] == 0:
                new_g = g_cost[current] + 1
                
                if neighbor not in g_cost or new_g < g_cost[neighbor]:
                    g_cost[neighbor] = new_g
                    f_cost = new_g + heuristic(neighbor, goal)
                    
                    heapq.heappush(open_list, (f_cost, neighbor))
                    parent[neighbor] = current
                    
    return None
    
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of cols: "))

grid = []

print("Enter grid (0 = free, 1 = obstacle): ")
for i in range(rows):
    row = list(map (int, input().split()))
    grid.append(row)

# grid = [[0, 0, 0, 0],
# [1, 1, 0, 1],
# [0, 0, 0, 0],
# [0, 1, 1, 0]]
    
start = tuple(map (int, input("Enter start (row, col): ").split()))
goal = tuple(map (int, input("Enter goal (row, col): ").split()))
# start = (0, 0)
# goal = (3, 3)

path = a_star_algorithm(start, goal, grid)

if path: 
    print("\nShortest Path:", path)
else:
    print("\nNo path found")
    
```

```python

# Prim's Algorithm

import heapq
from collections import defaultdict

def primsAlgorithm(graph, start):
    minHeap = [(0, start)]   # (cost, node)
    visited = set()
    totalCost = 0

    while minHeap:
        cost, node = heapq.heappop(minHeap)

        if node in visited:
            continue

        visited.add(node)
        totalCost += cost

        for neighbor, weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(minHeap, (weight, neighbor))

    return totalCost


# ---------------- DRIVER CODE ----------------
graph = defaultdict(list)

# graph = {
#     0: [(1, 2), (3, 6)],
#     1: [(0, 2), (2, 3), (3, 8), (4, 5)],
#     2: [(1, 3), (4, 7)],
#     3: [(0, 6), (1, 8)],
#     4: [(1, 5), (2, 7)]
# }

# Input
n = int(input("Enter number of edges: "))
print("Enter edges (u v weight):")

for _ in range(n):
    u, v, w = map(int, input().split())

    # Undirected graph
    graph[u].append((v, w))
    graph[v].append((u, w))

start = int(input("Enter starting node: "))

# Output
print("Prim's MST Cost:", primsAlgorithm(graph, start))

# Enter number of edges: 7
# Enter edges (u v weight):
# 0 1 2
# 0 3 6
# 1 2 3
# 1 3 8
# 1 4 5
# 2 4 7
# 3 4 9
# Enter starting node: 0
```

```python
# Kruskal's Algorithm with user input (same style as BFS/DFS and Prim's)

def find(parent, i):
    if parent[i] != i:
        parent[i] = find(parent, parent[i])   # Path Compression
    return parent[i]


def union(parent, rank, x, y):
    rootX = find(parent, x)
    rootY = find(parent, y)

    if rootX != rootY:
        if rank[rootX] > rank[rootY]:
            parent[rootY] = rootX
        elif rank[rootX] < rank[rootY]:
            parent[rootX] = rootY
        else:
            parent[rootY] = rootX
            rank[rootX] += 1


def kruskalsMst(edges, n):
    # Sort edges by weight
    edges.sort(key=lambda x: x[2])

    parent = list(range(n))
    rank = [0] * n
    mst_cost = 0

    for u, v, w in edges:
        if find(parent, u) != find(parent, v):
            union(parent, rank, u, v)
            mst_cost += w

    return mst_cost

# Enter number of vertices: 5
# Enter number of edges: 6
# Enter edges (u v weight):
# 0 1 2
# 0 3 6
# 1 2 3
# 1 3 8
# 1 4 5
# 2 4 7

# ---------------- DRIVER CODE ----------------
n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

edges = []

print("Enter edges (u v weight):")
for _ in range(e):
    u, v, w = map(int, input().split())
    edges.append((u, v, w))

# Output
print("Kruskal's MST Cost:", kruskalsMst(edges, n))
```

```python
# N Queen's

def solve_n_queens(n):
    solutions = []
    cols = set()
    pos_diag = set()
    neg_diag = set()

    board = [["."] * n for _ in range(n)]

    def backtrack(r):
        if r == n:
            copy = [" ".join(row) for row in board]
            solutions.append(copy)
            return

        for c in range(n):
            if c in cols or (r + c) in pos_diag or (r - c) in neg_diag:
                continue

            cols.add(c)
            pos_diag.add(r + c)
            neg_diag.add(r - c)
            board[r][c] = "Q"

            backtrack(r + 1)

            cols.remove(c)
            pos_diag.remove(r + c)
            neg_diag.remove(r - c)
            board[r][c] = "."

    backtrack(0)
    return solutions


result = solve_n_queens(4)

print(f"Solutions: {len(result)}")

if len(result) == 0:
    print("No solution")
else:
    print(f"Number of solutions: {len(result)}")

for idx, sol in enumerate(result, 1):
    print(f"Solution {idx}:")
    for row in sol:
        print(row)
    print()
```
